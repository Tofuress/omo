```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    'y': [8.5, 12, 9.2, 14.5, 10.8, 7.2, 11.9, 15.2, 8.9, 11.5, 7.8, 13.2, 9, 14], # цена
    'x1': [32, 45, 38, 52, 41, 29, 47, 55, 34, 44, 31, 49, 36, 50],# площадь
    'x2': [1, 5, 3, 8, 2, 1, 6, 9, 4, 5, 1, 7, 2, 8], # этаж
})

X = df[['x1', 'x2']]
y = df['y']

model = LinearRegression()
model.fit(X, y)

print(f"Уравнение: y = {model.intercept_:.2f} + {model.coef_[0]:.2f}*x1 + {model.coef_[1]:.2f}*x2")

price_house = np.array([[30, 8]])
pred = model.predict(price_house)
print(f"Прогноз для x1=30, x2=8: {pred[0]:.2f} млн руб.")

pred2 = model.predict(X)
print(f"R2: {r2_score(y, pred2):.3f}")
print(f"MSE: {mean_squared_error(y, pred2):.1f} млн руб.")
```
